```python
# 16.4

import sqlite3

# Create/connect to books.db
conn = sqlite3.connect("books.db")

# Create a cursor
cursor = conn.cursor()

# Create the books table
cursor.execute("""
CREATE TABLE IF NOT EXISTS books (
    title TEXT,
    author TEXT,
    year INTEGER
)
""")

# Save changes
conn.commit()

# Close database
conn.close()

```
